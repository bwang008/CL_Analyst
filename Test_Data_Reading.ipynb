```python
import pandas as pd
import pandas_ta as ta
import numpy as np
import yfinance as yf
```

```python
#data = pd.read_csv('data/cl-5m_bk.csv',sep=';',parse_dates=[[0,1]],index_col=0,dayfirst=True)
data = pd.read_csv('data/test.csv',sep=';',header=None,index_col=None)
        
cols=['Day','Time','Open','High','Low','Close','Volume']
data.columns = cols
```

```python
data.head()
```

```python
data['Date'] = pd.to_datetime(data['Day'] + ' ' + data['Time'],
                                 format='%d/%m/%Y %H:%M')
data = data.set_index('Date')
```

```python
#Drop the DT columns
data = data.drop(['Day','Time'],axis=1)
```

```python

ticker = "aapl"
# Download data directly using yfinance
# Adjust period, start/end dates as needed
aapl_price = yf.download(ticker, period="5y")

if data is not None and not data.empty:

    aapl_price.columns = data.columns.get_level_values(0)
    print("Data downloaded successfully:")
    print(aapl_price.head())

    # Now, apply pandas_ta indicators to this DataFrame
    # Example: Add Simple Moving Average
    aapl_price.ta.sma(length=50, append=True)

    # Example: Add Relative Strength Index
    aapl_price.ta.rsi(append=True)

    # Example: Add MACD
    aapl_price.ta.macd(append=True)

    print("\nData with indicators:")
    print(aapl_price.head())
    # Assign the result back to df if needed
    df = aapl_price
else:
    print(f"Failed to download data for {ticker}")
```
