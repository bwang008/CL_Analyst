```python
cols=['Day','Time','Open','High','Low','Close','Volume']
cl = pd.read_csv('data/test.csv', sep=';', index_col=None, header=None, names=cols) #initial dataset 100 records to get basic functionality working.
#cl = pd.read_csv('data/test10k.csv') #larger dataset to test with 

# Create date column by combining Day + Time. Set it as the index.
cl['Date'] = pd.to_datetime(cl['Day'] + ' ' + cl['Time'],
                             format='%d/%m/%Y %H:%M')
cl = cl.set_index('Date')

#Drop Day + Time
cl = cl.drop(['Day','Time'],axis=1)
```

```python
# Add indicators to the DataFrame
cl['sma_10'] = ta.sma(cl['Close'], length=10)  # Simple Moving Average
cl['ema_10'] = ta.ema(cl['Close'], length=10)  # Exponential Moving Average
cl['rsi'] = ta.rsi(cl['Close'])               # Relative Strength Index
cl['macd'], cl['macd_signal'], cl['macd_hist'] = ta.macd(cl['Close'])  # MACD
cl['stoch_k'], cl['stoch_d'] = ta.stoch(cl['High'], cl['Low'], cl['Close'])  # Stochastic Oscillator
#cl['bb_upper'], cl['bb_middle'], cl['bb_lower'] = ta.bbands(cl['Close'])  # Bollinger Bands
#cl['adx'] = ta.adx(cl['High'], cl['Low'], cl['Close'])  # Average Directional Index
cl['atr'] = ta.atr(cl['High'], cl['Low'], cl['Close'])  # Average True Range
#cl['cci'] = ta.cci(cl['High'], cl['Low'], cl['Close'], cl['Close'])  # Commodity Channel Index
cl['roc'] = ta.roc(cl['Close'])  # Rate of Change
cl['williams_r'] = ta.willr(cl['High'], cl['Low'], cl['Close'])  # Williams %R
cl['obv'] = ta.obv(cl['Close'], cl['Volume'])  # On-Balance Volume
cl['mfi'] = ta.mfi(cl['High'], cl['Low'], cl['Close'], cl['Volume'])  # Money Flow Index
#cl['psar'] = ta.psar(cl['High'], cl['Low'], cl['Close'])  # Parabolic SAR
cl['vwap'] = ta.vwap(cl['High'], cl['Low'], cl['Close'], cl['Volume'])  # Volume Weighted Average Price
#cl['tsi'] = ta.tsi(cl['Close'])  # True Strength Index
cl['uo'] = ta.uo(cl['High'], cl['Low'], cl['Close'])  # Ultimate Oscillator
cl['kama'] = ta.kama(cl['Close'])  # Kaufman's Adaptive Moving Average
cl['dpo'] = ta.dpo(cl['Close'])  # Detrended Price Oscillator
#cl['ppo'] = ta.ppo(cl['Close'])  # Percentage Price Oscillator
cl['mom'] = ta.mom(cl['Close'])  # Momentum
cl['trix'] = ta.trix(cl['Close'])  # TRIX
cl['natr'] = ta.natr(cl['High'], cl['Low'], cl['Close'])  # Normalized ATR
cl['donchian_upper'], cl['donchian_lower'] = ta.donchian(cl['High'], cl['Low'], cl['Close'])  # Donchian Channels
cl['stdev'] = ta.stdev(cl['Close'])  # Standard Deviation
cl['var'] = ta.var(cl['Close'])  # Variance
cl['skew'] = ta.skew(cl['Close'])  # Skewness
cl['kurt'] = ta.kurt(cl['Close'])  # Kurtosis
cl['zscore'] = ta.zscore(cl['Close'])  # Z-Score
cl['entropy'] = ta.entropy(cl['Close'])  # Entropy
cl['pvol'] = ta.pvol(cl['Close'], cl['Volume'])  # Price Volume Trend
cl['vortex_pos'], cl['vortex_neg'] = ta.vortex(cl['High'], cl['Low'], cl['Close'])  # Vortex Indicator
cl['tsi'] = ta.tsi(cl['Close'])  # True Strength Index
cl['adx_pos'], cl['adx_neg'] = ta.adx(cl['High'], cl['Low'], cl['Close'])  # ADX Positive/Negative
cl['aroon_up'], cl['aroon_down'] = ta.aroon(cl['Close'])  # Aroon Indicator
cl['cmf'] = ta.cmf(cl['High'], cl['Low'], cl['Close'], cl['Volume'])  # Chaikin Money Flow
cl['efi'] = ta.efi(cl['Close'], cl['Volume'])  # Elder's Force Index
cl['pvi'] = ta.pvi(cl['Close'], cl['Volume'])  # Positive Volume Index
cl['nvi'] = ta.nvi(cl['Close'], cl['Volume'])  # Negative Volume Index
#cl['tsi'] = ta.tsi(cl['Close'])  # True Strength Index
cl['rvi'] = ta.rvi(cl['High'], cl['Low'], cl['Close'])  # Relative Volatility Index
cl['cmo'] = ta.cmo(cl['Close'])  # Chande Momentum Oscillator
cl['qstick'] = ta.qstick(cl['Open'], cl['Close'])  # QStick Indicator
cl['fisher'] = ta.fisher(cl['High'], cl['Low'])  # Fisher Transform
cl['hma'] = ta.hma(cl['Close'])  # Hull Moving Average
cl['linreg'] = ta.linreg(cl['Close'])  # Linear Regression
cl['sinwma'] = ta.sinwma(cl['Close'])  # Sine Weighted Moving Average

print("Indicators generated successfully.")
return cl
```
